```python
# import rich for formatting output
from rich import print as rprint 

# initialize matrix 1
m1 = [
    ([1, 2, 3]),
    ([2, 3, 4]),
    ([4, 5, 8]),
]

# initialize matrix 2
m2 = [
    ([2, 7, 4]),
    ([1, 1, 2]),
    ([3, 3, 3])
]

# initialization of resultant matrix
result = [
    ([0, 0, 0]),
    ([0, 0, 0]),
    ([0, 0, 0])
]

# loops for addtion
for i in range(len(m1)):
    for j in range(len(m1[0])):
            result[i][j] = m1[i][j] + m2[i][j]

# print results
rprint(f"First Matrix:  [bright_cyan]{m1}[/bright_cyan]")
rprint(f"Second Matrix: [bright_green]{m2}[/bright_green]")
rprint(f"Result:        [bright_yellow]{result}[/bright_yellow]")
```
