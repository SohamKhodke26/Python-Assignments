```python
import re

# set up pattern to match and text to check
pattern = r"[a-zA-Z0-9]+"
text1 = "Example2"
text2 = "ExamPLE_2."

# typecast into boolean data type
t1 = bool(re.fullmatch(pattern, text1))
t2 = bool(re.fullmatch(pattern, text2))

# print result
print(t1, t2)
```
