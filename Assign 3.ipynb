```python
#input
sides = int(input("Enter number of sides: "))

if sides == 3:
    print("The shape is a Triangle")

elif sides == 4:
     lenght = float(input("Enter lenght: "))
     width = float(input("Enter width: "))

     if lenght == width:
         print("The shape is Square")
     else:
         print("The shape is Rectangle")

else:
    print("Unknown shape")
```
